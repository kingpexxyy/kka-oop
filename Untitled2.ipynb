```python
class Kucing:
# Constructor
  def __init__(self, nama, warna):
    self.nama = nama # Attribute
    self.warna = warna # Attribute
# Method
  def suara(self):
    return f"{self.nama} berkata: Meow!"
# Membuat Object (Instansiasi)
kucing1 = Kucing("Luna", "Putih")
kucing2 = Kucing("Garfield", "Oranye")
# Mengakses Attribute dan Method
print(kucing1.nama) # Output: Luna
print(kucing2.suara()) # Output: Garfield berkata: Meow!
```

```python
class AkunBank:
  def __init__(self, saldo):
    self.__saldo = saldo # Private attribute

  def cek_saldo(self):
    return self.__saldo

  def setor(self, jumlah):
    if jumlah > 0:
      self.__saldo += jumlah
      print(f"Berhasil menyetor {jumlah:.2f}. Saldo saat ini: {self.__saldo:.2f}")
    else:
      print("Jumlah setor harus positif.")

  def tarik(self, jumlah):
    if jumlah > 0:
      if self.__saldo >= jumlah:
        self.__saldo -= jumlah
        print(f"Berhasil menarik {jumlah:.2f}. Saldo saat ini: {self.__saldo:.2f}")
      else:
        print("Saldo tidak mencukupi.")
    else:
      print("Jumlah tarik harus positif.")

rekening = AkunBank(1000)

while True:
  print("\n--- Menu Transaksi Bank ---")
  print("1. Setor Tunai")
  print("2. Tarik Tunai")
  print("3. Cek Saldo")
  print("4. Keluar")
  pilihan = input("Masukkan pilihan Anda (1-4): ")

  if pilihan == '1':
    try:
      jumlah = float(input("Masukkan jumlah untuk disetor: "))
      rekening.setor(jumlah)
    except ValueError:
      print("Input tidak valid. Harap masukkan angka.")
  elif pilihan == '2':
    try:
      jumlah = float(input("Masukkan jumlah untuk ditarik: "))
      rekening.tarik(jumlah)
    except ValueError:
      print("Input tidak valid. Harap masukkan angka.")
  elif pilihan == '3':
    print(f"Saldo Anda saat ini: {rekening.cek_saldo():.2f}")
  elif pilihan == '4':
    print("Terima kasih telah menggunakan layanan kami.")
    break
  else:
    print("Pilihan tidak valid. Silakan coba lagi.")
```
